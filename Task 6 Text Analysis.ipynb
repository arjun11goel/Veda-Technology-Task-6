# Text Analysis Project
**Objective:** Analyze a given text to calculate the number of words, characters, vowels, and repeated words to practice basic natural language processing (NLP) techniques.

```python
# Import necessary libraries
import re
from collections import Counter

# Define a sample paragraph dataset
text = """
Natural language processing (NLP) is a subfield of linguistics, computer science, 
and artificial intelligence. It focuses on the interactions between computers and human language. 
The ultimate objective of NLP is to read, decipher, understand, and make sense of the human languages 
in a manner that is valuable. Language processing is a very interesting field of study.
"""

print("Original Text:\n", text.strip())
```

## 1. Character and Word Counts
Calculating the total number of characters (including spaces) and the total number of words using `len()` and `split()`.

```python
# Character Count
char_count = len(text)

# Word Count
# We use split() which splits by whitespace
words = text.split()
word_count = len(words)

print(f"Total Characters: {char_count}")
print(f"Total Words: {word_count}")
```

## 2. Vowel Analysis
Counting the number of vowels (A, E, I, O, U) in the text.

```python
# Vowel Count
vowels = set("aeiou")
vowel_count = 0

# Convert text to lowercase to ensure we catch both uppercase and lowercase vowels
for char in text.lower():
    if char in vowels:
        vowel_count += 1
        
print(f"Total Vowels: {vowel_count}")
```

## 3. Repeated Words Analysis
Using a dictionary/Counter to identify which words appear more than once. We will first clean the text to remove punctuation and make it lowercase so that "Language" and "language." are treated as the same word.

```python
# Clean the text: convert to lowercase and extract only alphabetic words using regex
clean_words = re.findall(r'\b[a-z]+\b', text.lower())

# Use Counter to count word frequencies
word_frequencies = Counter(clean_words)

# Filter out words that only appear once
repeated_words = {word: count for word, count in word_frequencies.items() if count > 1}

print("Repeated Words:")
for word, count in repeated_words.items():
    print(f"- '{word}': {count} times")
```

## Interview Questions & Answers

**1. What is text preprocessing?**
Text preprocessing is the practice of cleaning and preparing raw text data into a format that a machine learning model or NLP algorithm can process and understand. Real-world text is often messy, so preprocessing typically involves steps such as:
* Converting all text to lowercase to ensure uniformity.
* Removing punctuation, special characters, and numbers.
* Tokenization (breaking text down into smaller units like words or sentences).
* Removing "stop words" (common words like "the", "is", "in" that add little meaning).
* Stemming or lemmatization (reducing words to their base or root form).

**2. How can repeated words be identified?**
Repeated words are best identified by first standardizing the text (lowercasing and removing punctuation) so that variations of the same word are treated identically. Then, you can tokenize the text into a list of words. To find repetitions, you can iterate through the list and count occurrences using a hash map (like a standard Python `dict`) or utilize specialized tools like the `Counter` class from Python's `collections` module. Any word with a frequency count greater than 1 is a repeated word. 

**3. What is NLP?**
NLP, or Natural Language Processing, is a branch of artificial intelligence and computer science focused on the interaction between computers and human language. Its goal is to give computers the ability to read, understand, interpret, and generate human language in a way that is meaningful and useful. Common applications of NLP include translation software, chatbots, sentiment analysis, and voice assistants.
